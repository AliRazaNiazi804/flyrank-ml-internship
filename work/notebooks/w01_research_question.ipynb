## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

### My Lane: Predicting Customer Churn in a Subscription Service

I've chosen to focus on predicting customer churn in a hypothetical subscription service. This lane is critical because customer retention is often more cost-effective than customer acquisition. By identifying customers at risk of churning, the business can proactively intervene with targeted strategies to improve satisfaction and loyalty.

```python
import pandas as pd
import numpy as np

# Simulate a hypothetical customer dataset for churn prediction
np.random.seed(42) # for reproducibility
num_customers = 1000

data = {
    'customer_id': np.arange(1, num_customers + 1),
    'subscription_duration_months': np.random.randint(1, 60, num_customers),
    'monthly_spend': np.random.uniform(20, 150, num_customers),
    'last_login_days_ago': np.random.randint(1, 90, num_customers),
    'support_tickets_opened': np.random.randint(0, 10, num_customers),
    'churned': np.random.choice([0, 1], num_customers, p=[0.85, 0.15]) # 15% churn rate
}

customer_df = pd.DataFrame(data)

print("--- Simulated Customer Data ---")
display(customer_df.head())

# Calculate a provisional churn rate as a key metric for the lane
churn_rate = customer_df['churned'].mean() * 100
print(f"\nProvisional Churn Rate: {churn_rate:.2f}%")
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### The Decision: When to Intervene with At-Risk Customers

My work will improve the decision of *when* and *how* to intervene with customers who are likely to churn. Marketing and customer success teams will act on this by implementing personalized outreach, special offers, or tailored support. A wrong recommendation (e.g., marking a loyal customer as at-risk, or missing a truly at-risk customer) could cost the business in wasted marketing spend, decreased customer lifetime value, or ultimately, lost revenue.

```python
import numpy as np

# Assuming an average customer lifetime value (CLV)
avg_clv = 500  # dollars

# Assuming cost of customer acquisition (CAC)
cac = 100 # dollars

# Cost of a false negative (missed churner): lost CLV
false_negative_cost = avg_clv

# Cost of a false positive (customer incorrectly flagged as churn risk):
# e.g., cost of an unnecessary retention offer
false_positive_cost = 20 # dollars for a discount or special outreach

print(f"Average Customer Lifetime Value (CLV): ${avg_clv}")
print(f"Cost of Customer Acquisition (CAC): ${cac}")
print(f"Cost of a False Negative (missed churner) = Lost CLV: ${false_negative_cost}")
print(f"Cost of a False Positive (unnecessary intervention): ${false_positive_cost}")

# Illustrate a scenario with a hypothetical number of errors
hypothetical_false_negatives = 50
hypothetical_false_positives = 200

total_cost_of_wrong_calls = (hypothetical_false_negatives * false_negative_cost) + \
                           (hypothetical_false_positives * false_positive_cost)

print(f"\nHypothetical scenario with {hypothetical_false_negatives} false negatives and {hypothetical_false_positives} false positives:")
print(f"Total estimated cost of wrong calls: ${total_cost_of_wrong_calls}")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### Careful Words: What I Can and Can't Claim

My work will be able to provide **decision-support** by highlighting customer segments with a higher propensity to churn, allowing for **directional** interventions. The models will show **observed** relationships between customer behavior (e.g., login frequency, feature usage) and churn. I can claim that certain features are *correlated* with churn, but I **cannot claim causal proof** that changing a specific feature will definitively prevent churn. Furthermore, I will not be 'predicting Google' or making claims about market-wide behavior beyond the scope of this specific dataset.

```python
import pandas as pd
import numpy as np # Added for re-simulation if customer_df is missing

# Check if customer_df is already defined, otherwise create a placeholder or re-simulate
# This makes the cell runnable even if previous cells were not executed in order.
if 'customer_df' not in locals() and 'customer_df' not in globals():
    print("Warning: 'customer_df' not found. Re-simulating a basic dataset for demonstration.")
    np.random.seed(42) # for reproducibility
    num_customers = 1000
    data = {
        'customer_id': np.arange(1, num_customers + 1),
        'subscription_duration_months': np.random.randint(1, 60, num_customers),
        'monthly_spend': np.random.uniform(20, 150, num_customers),
        'last_login_days_ago': np.random.randint(1, 90, num_customers),
        'support_tickets_opened': np.random.randint(0, 10, num_customers),
        'churned': np.random.choice([0, 1], num_customers, p=[0.85, 0.15]) # 15% churn rate
    }
    customer_df = pd.DataFrame(data)
    print("Please run the cell 'amgBLeTnxcEw' first for the intended 'customer_df'.")

# Use the previously simulated customer_df
# Assuming customer_df is available from cell amgBLeTnxcEw

# Observe the relationship between subscription duration and churned status
# This shows a directional relationship without claiming causality
churn_by_duration = customer_df.groupby('subscription_duration_months')['churned'].mean().reset_index()

print("--- Average Churn Rate by Subscription Duration (first 10 unique durations) ---")
display(churn_by_duration.head(10))

# Observe the relationship between support tickets and churned status
churn_by_tickets = customer_df.groupby('support_tickets_opened')['churned'].mean().reset_index()

print("\n--- Average Churn Rate by Number of Support Tickets Opened ---")
display(churn_by_tickets)

# Calculate a simple correlation (observed relationship)
correlation_duration_churn = customer_df['subscription_duration_months'].corr(customer_df['churned'])
correlation_tickets_churn = customer_df['support_tickets_opened'].corr(customer_df['churned'])

print(f"\nCorrelation between Subscription Duration and Churn: {correlation_duration_churn:.2f}")
print(f"Correlation between Support Tickets and Churn: {correlation_tickets_churn:.2f}")
```
